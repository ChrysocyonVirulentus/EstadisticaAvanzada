📝 **Consigna 7:** ¿Cómo harían si tuvieran agregar una segunda tabla que comparte las mismas columnas? ¿Y si comparten algunas filas?

````python
df_nuevo = {
    "PDB ID" : "36CW",
    "Chain ID" : "A",
    "Structure Title" : "HIV-1 reverse transcriptase (F227C) in complex with DNAddG Aptamer and unincorporated ISL-triphosphate ",
    "Resolution" :  1.90,
    "Ligand ID" : "NaN",
    "Ligand Name" : "NaN",
    "Macromolecule Name" : "NaN",
    "doble_resolucion" : "NaN",
    "col1+col2" : "NaN",
    "Taxonomy ID" : 11676
}

#Concatenar
datos_pdb = pd.concat([datos_pdb,pd.DataFrame(df_nuevo,index=[0])],ignore_index=True)
datos_pdb
````

````python
# Eliminar la ultima fila
datos_pdb = datos_pdb.drop(datos_pdb.index[-1])
datos_pdb
````

````python
# Sacar las filas de 2 a 100 en un nuevo data frame
df_nuevo_2 = datos_pdb.iloc[2:100]

# Concatenar a df_nuevo
df_nuevo_2 = pd.concat([df_nuevo_2,pd.DataFrame(df_nuevo,index=[0])],ignore_index=True)

# Unirlo a la tabla original

datos_pdb = pd.concat([datos_pdb,df_nuevo_2],ignore_index=True)

# Eliminar duplicados
datos_pdb = datos_pdb.drop_duplicates()

# Corregir indice
datos_pdb = datos_pdb.reset_index(drop=True)

datos_pdb
````

````python
# Transformar la Columna Molecular name en categorico
datos_pdb['Macromolecule Name'] = datos_pdb['Macromolecule Name'].dropna().astype('category')

# Construir una serie con esos categoricos
categorias = pd.Series(datos_pdb['Macromolecule Name'].cat.categories)

# Imprimir la serie
print(categorias)
````
